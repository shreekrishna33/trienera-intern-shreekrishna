```python

import pandas as pd

# Load the agriculture AI tools we researched
df = pd.read_csv("agri_ai_tools.csv")


# 1. Count the total number of tools
print("=" * 45)
print("AGRICULTURE AI TOOLS ANALYSIS")
print("=" * 45)

print("\n1. Total number of tools researched:")
print(len(df))


# 2. Display all the tool names
print("\n2. Names of the AI tools:")

for tool in df["Tool Name"]:
    print("-", tool)


# 3. Find the number of tools in each category
print("\n3. Number of tools in each category:")
print(df["Category"].value_counts().to_string())


# 4. Check which tools mention Kannada support
print("\n4. Tools mentioning Kannada support:")

kannada_tools = df[
    df["Language Support"].str.contains(
        "Kannada", case=False, na=False
    )
]

if len(kannada_tools) > 0:
    print(kannada_tools["Tool Name"].to_string(index=False))
else:
    print("No tools currently mention Kannada support.")

print("Total tools mentioning Kannada:", len(kannada_tools))


# 5. Check for empty cells in the dataset
print("\n5. Checking for missing information:")

missing_info = df.isnull().sum()

if missing_info.sum() == 0:
    print("No empty cells found in the dataset.")
else:
    print(missing_info[missing_info > 0])


# 6. Find information that still needs research
print("\n6. Details that need further research:")

found_pending_details = False

for column in df.columns:
    pending = df[column].astype(str).str.contains(
        "verify|research", case=False, na=False
    )

    if pending.any():
        found_pending_details = True
        print(f"\nColumn: {column}")
        print(
            df.loc[pending, ["Tool Name", column]]
            .to_string(index=False)
        )

if not found_pending_details:
    print("No entries marked for verification were found.")


# 7. Show a quick summary of the dataset
print("\n7. Dataset summary:")

print("Number of tools:", len(df))
print("Number of columns:", len(df.columns))
print("Column names:", ", ".join(df.columns))


# 8. Save a clean copy of the analyzed dataset
df.to_csv("agri_ai_tools_analyzed.csv", index=False)

print("\nAnalysis completed successfully!")
print("Analyzed dataset saved as agri_ai_tools_analyzed.csv")
```
