```python

import pandas as pd
from pathlib import Path

# ==========================================
# AGRICULTURE AI TOOLS ANALYSIS
# ==========================================

print("=" * 45)
print("AGRICULTURE AI TOOLS ANALYSIS")
print("=" * 45)

# 1. Load our research data from the CSV file
file_name = "agri_ai_tools.csv"

if not Path(file_name).exists():
    raise FileNotFoundError(
        f"Cannot find {file_name}. "
        "Please keep it in your project folder."
    )

df = pd.read_csv(file_name)

# Clean spaces around column names
df.columns = df.columns.str.strip()

# Check that the important columns are available
required_columns = [
    "Tool Name",
    "Category",
    "Function",
    "AI/Technology Used",
    "Language Support",
    "Strength",
    "Weakness",
    "Source URL"
]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(
        "Missing columns: " + ", ".join(missing_columns)
    )

# Remove completely empty rows
df = df.dropna(how="all").copy()

# ==========================================
# 2. COUNT THE TOOLS
# ==========================================

print("\n1. Total number of tools researched:")
print(len(df))

# ==========================================
# 3. DISPLAY THE TOOL NAMES
# ==========================================

print("\n2. Names of the AI tools:")

for number, tool in enumerate(df["Tool Name"], start=1):
    print(f"{number}. {tool}")

# ==========================================
# 4. COUNT TOOLS IN EACH CATEGORY
# ==========================================

print("\n3. Number of tools in each category:")

print(df["Category"].value_counts().to_string())

# ==========================================
# 5. CHECK FOR KANNADA LANGUAGE MENTIONS
# ==========================================

print("\n4. Tools mentioning Kannada support:")

kannada_tools = df[
    df["Language Support"].fillna("").str.contains(
        "Kannada",
        case=False,
        na=False
    )
]

if not kannada_tools.empty:
    print(
        kannada_tools["Tool Name"].to_string(index=False)
    )
else:
    print("No tools mention Kannada in the dataset.")

print("Total tools mentioning Kannada:", len(kannada_tools))

# Note: This checks the CSV text, not the actual language
# capabilities of each application.

# ==========================================
# 6. CHECK FOR MISSING INFORMATION
# ==========================================

print("\n5. Checking for missing information:")

missing_info = df[required_columns].isna().sum()

if missing_info.sum() == 0:
    print("No empty cells found in the required columns.")
else:
    print(missing_info[missing_info > 0].to_string())

# Also check for cells that contain only spaces
blank_info = (
    df[required_columns]
    .fillna("")
    .apply(lambda column: column.astype(str).str.strip().eq(""))
    .sum()
)

if blank_info.sum() > 0:
    print("\nColumns containing blank text:")
    print(blank_info[blank_info > 0].to_string())

# ==========================================
# 7. FIND DETAILS THAT NEED MORE RESEARCH
# ==========================================

print("\n6. Details that may need further research:")

pending_words = (
    r"verify|to be researched|needs further investigation|"
    r"need further investigation|needs investigation"
)

found_pending_details = False

for column in required_columns:
    pending = df[column].fillna("").astype(str).str.contains(
        pending_words,
        case=False,
        regex=True,
        na=False
    )

    if pending.any():
        found_pending_details = True
        print(f"\nColumn: {column}")

        print(
            df.loc[pending, ["Tool Name", column]]
            .to_string(index=False)
        )

if not found_pending_details:
    print("No obvious research reminders were found.")

# ==========================================
# 8. SHOW A DATASET SUMMARY
# ==========================================

print("\n7. Dataset summary:")

print("Total tools:", len(df))
print("Total columns:", len(df.columns))
print("Different categories:", df["Category"].nunique())

# ==========================================
# 9. SAVE THE UPDATED DATASET
# ==========================================

output_file = "agri_ai_tools_analyzed.csv"

df.to_csv(output_file, index=False, encoding="utf-8-sig")

print("\n8. Output file:")
print("Saved the analyzed dataset as:", output_file)

print("\n" + "=" * 45)
print("Analysis completed successfully!")
print("=" * 45)
```
